### Hue-Saturation-Value Model

Use `from matplotlib.colors import hsv_to_rgb, rgb_to_hsv`

```python
from matplotlib.colors import hsv_to_rgb, rgb_to_hsv
```

```python
# RBG -> H, S, V ou HSV
```

```python
# rgb_to_hsv(t2f)
```

```python
hsv = rgb_to_hsv(t2f.permute(1, 2, 0))
```

```python
hsv.shape
```

```python
hue = hsv[:,:,0]
hue.min(), hue.max()
```

```python
plt.imshow(hue, cmap="hsv")
plt.colorbar()
```

```python
hue2 = (hue + 0.5) % 1.0
hue2 = torch.tensor(hue2)
```

```python
plt.imshow(hue2, cmap="hsv")
plt.colorbar()
```

```python
# hsv[:,:,0] = hue2 # alternative
```

```python
sat = torch.tensor(hsv[:,:,1])
val = torch.tensor(hsv[:,:,2])
hsv = torch.stack((hue2, sat, val), dim=2)
hsv.shape
```

```python
rgb = torch.tensor(hsv_to_rgb(hsv))
rgb.shape
```

```python
tensor_to_image(rgb.permute(2, 0 , 1))
```
